```python
import pandas as pd

# URL para o arquivo bruto (raw) no GitHub
url = 'https://raw.githubusercontent.com/Sainathmudigonda/Student-Result-Analysis/main/student_exam_scores.csv'

# Carregando os dados
df = pd.read_csv(url)

# Exibindo as primeiras linhas do DataFrame
display(df.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# Remover valores nulos para o cálculo da regressão
df_clean = df[['hours_studied', 'exam_score']].dropna()
x = df_clean['hours_studied']
y = df_clean['exam_score']

# Calcula a regressão linear
slope, intercept, r_value, p_value, std_err = stats.linregress(x, y)
r_squared = r_value ** 2

# Configura o estilo visual
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 6))

# Desenha o gráfico de dispersão com a reta de regressão
sns.regplot(
    data=df_clean,
    x='hours_studied',
    y='exam_score',
    scatter_kws={'alpha': 0.7, 'color': '#1f77b4'},
    line_kws={'color': 'red', 'linewidth': 2}
)

# Cria o texto com a equação e o R²
texto_regressao = f'y = {slope:.2f}x + {intercept:.2f}\n$R^2$ = {r_squared:.4f}'

# Adiciona o texto no gráfico (ajustando a posição x e y conforme os limites dos seus dados)
plt.text(
    x=x.min() + 0.5,
    y=y.max() - 10,
    s=texto_regressao,
    fontsize=12,
    color='red',
    bbox=dict(facecolor='white', alpha=0.8, edgecolor='gray', boxstyle='round,pad=0.5')
)

# Títulos e legendas
plt.title('Relação entre Horas de Estudo e Nota da Prova (com Regressão)', fontsize=14, pad=15)
plt.xlabel('Horas de Estudo', fontsize=12)
plt.ylabel('Nota da Prova', fontsize=12)

# Exibe o gráfico
```
