```python
import pandas as pd

# URL para o arquivo bruto (raw) no GitHub
url = 'https://raw.githubusercontent.com/Sainathmudigonda/Student-Result-Analysis/main/student_exam_scores.csv'

# Carregando os dados
df = pd.read_csv(url)

# Exibindo as primeiras linhas do DataFrame
display(df.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Configura o estilo visual
sns.set_theme(style="whitegrid")

# Cria a figura
plt.figure(figsize=(10, 6))

# Desenha o gráfico de dispersão com a reta de regressão linear
sns.regplot(
    data=df,
    x='hours_studied',
    y='exam_score',
    scatter_kws={'alpha': 0.7, 'color': '#1f77b4'},
    line_kws={'color': 'red', 'linewidth': 2}
)

# Títulos e legendas
plt.title('Relação entre Horas de Estudo e Nota da Prova', fontsize=14, pad=15)
plt.xlabel('Horas de Estudo', fontsize=12)
plt.ylabel('Nota da Prova', fontsize=12)

# Exibe o gráfico
plt.show()
```
